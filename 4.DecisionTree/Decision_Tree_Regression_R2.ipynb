```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score
```

```python
dataset = pd.read_csv('50_Startups.csv')

dataset = pd.get_dummies(dataset, drop_first=True)

dataset.head()
```

```python
indep = dataset[['R&D Spend',
                 'Administration',
                 'Marketing Spend',
                 'State_Florida',
                 'State_New York']]

dep = dataset[['Profit']]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    indep,
    dep,
    test_size=1/3,
    random_state=0
)
```

```python
regressor = DecisionTreeRegressor(random_state=0)

regressor.fit(X_train, y_train.values.ravel())

y_pred = regressor.predict(X_test)

r_score = r2_score(y_test, y_pred)

print("R2 Score:", r_score)
```

```python
criteria = ['squared_error', 'absolute_error', 'poisson']
max_features_values = [None, 'sqrt', 'log2']
splitter_values = ['best', 'random']

results = []

for criterion in criteria:
    for max_features in max_features_values:
        for splitter in splitter_values:

            regressor = DecisionTreeRegressor(
                criterion=criterion,
                max_features=max_features,
                splitter=splitter,
                random_state=0
            )

            regressor.fit(X_train, y_train.values.ravel())

            y_pred = regressor.predict(X_test)

            r_score = r2_score(y_test, y_pred)

            results.append({
                'Criterion': criterion,
                'Max Features': max_features,
                'Splitter': splitter,
                'R2': r_score
            })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by='R2',
    ascending=False
)

results_df
```

```python
## Additional Hyper Tuning

max_depth_values = [None, 2, 3, 5, 8, 10, 15]

max_depth_results = []

for max_depth_value in max_depth_values:

    regressor = DecisionTreeRegressor(
        criterion='absolute_error',
        max_features=None,
        splitter='best',
        max_depth=max_depth_value,
        random_state=0
    )

    regressor.fit(X_train, y_train.values.ravel())

    y_pred = regressor.predict(X_test)

    r_score = r2_score(y_test, y_pred)

    max_depth_results.append({
        'max_depth': max_depth_value,
        'R2': r_score
    })

max_depth_df = pd.DataFrame(max_depth_results)

max_depth_df = max_depth_df.sort_values(
    by='R2',
    ascending=False
)

max_depth_df
```

```python
min_samples_split_values = [2, 5, 10, 20]

min_samples_split_results = []

for min_samples_split_value in min_samples_split_values:

    regressor = DecisionTreeRegressor(
        criterion='absolute_error',
        max_features=None,
        splitter='best',
        max_depth=5,
        min_samples_split=min_samples_split_value,
        random_state=0
    )

    regressor.fit(X_train, y_train.values.ravel())

    y_pred = regressor.predict(X_test)

    r_score = r2_score(y_test, y_pred)

    min_samples_split_results.append({
        'min_samples_split': min_samples_split_value,
        'R2': r_score
    })

min_samples_split_df = pd.DataFrame(min_samples_split_results)

min_samples_split_df = min_samples_split_df.sort_values(
    by='R2',
    ascending=False
)

min_samples_split_df
```

```python
min_samples_leaf_values = [1, 2, 5, 10]

min_samples_leaf_results = []

for min_samples_leaf_value in min_samples_leaf_values:

    regressor = DecisionTreeRegressor(
        criterion='absolute_error',
        max_features=None,
        splitter='best',
        max_depth=5,
        min_samples_split=2,
        min_samples_leaf=min_samples_leaf_value,
        random_state=0
    )

    regressor.fit(X_train, y_train.values.ravel())

    y_pred = regressor.predict(X_test)

    r_score = r2_score(y_test, y_pred)

    min_samples_leaf_results.append({
        'min_samples_leaf': min_samples_leaf_value,
        'R2': r_score
    })

min_samples_leaf_df = pd.DataFrame(min_samples_leaf_results)

min_samples_leaf_df = min_samples_leaf_df.sort_values(
    by='R2',
    ascending=False
)

min_samples_leaf_df
```

```python
max_leaf_nodes_values = [None, 5, 10, 20, 50]

max_leaf_nodes_results = []

for max_leaf_nodes_value in max_leaf_nodes_values:

    regressor = DecisionTreeRegressor(
        criterion='absolute_error',
        max_features=None,
        splitter='best',
        max_depth=5,
        min_samples_split=2,
        min_samples_leaf=2,
        max_leaf_nodes=max_leaf_nodes_value,
        random_state=0
    )

    regressor.fit(X_train, y_train.values.ravel())

    y_pred = regressor.predict(X_test)

    r_score = r2_score(y_test, y_pred)

    max_leaf_nodes_results.append({
        'max_leaf_nodes': max_leaf_nodes_value,
        'R2': r_score
    })

max_leaf_nodes_df = pd.DataFrame(max_leaf_nodes_results)

max_leaf_nodes_df = max_leaf_nodes_df.sort_values(
    by='R2',
    ascending=False
)

max_leaf_nodes_df
```

```python
min_impurity_decrease_values = [0, 0.001, 0.01, 0.05]

min_impurity_results = []

for min_impurity_value in min_impurity_decrease_values:

    regressor = DecisionTreeRegressor(
        criterion='absolute_error',
        max_features=None,
        splitter='best',
        max_depth=5,
        min_samples_split=2,
        min_samples_leaf=2,
        max_leaf_nodes=None,
        min_impurity_decrease=min_impurity_value,
        random_state=0
    )

    regressor.fit(X_train, y_train.values.ravel())

    y_pred = regressor.predict(X_test)

    r_score = r2_score(y_test, y_pred)

    min_impurity_results.append({
        'min_impurity_decrease': min_impurity_value,
        'R2': r_score
    })

min_impurity_df = pd.DataFrame(min_impurity_results)

min_impurity_df = min_impurity_df.sort_values(
    by='R2',
    ascending=False
)

min_impurity_df
```
