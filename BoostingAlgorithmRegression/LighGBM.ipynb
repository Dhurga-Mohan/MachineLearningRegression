```python

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from lightgbm import LGBMRegressor

# 1. Load the dataset
data = pd.read_csv("insurance_pre.csv")

# 2. Convert categorical columns into numeric columns
data = pd.get_dummies(data, drop_first=True, dtype=int)

# 3. Separate input features and target
X = data[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
y = data['charges']

# 4. Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=0
)

# 5. Create the LightGBM model
model = LGBMRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=-1,
    random_state=0,
    verbosity=-1
)

# 6. Train the model
model.fit(X_train, y_train)

# 7. Predict insurance charges
predictions = model.predict(X_test)

# 8. Evaluate the model
score = r2_score(y_test, predictions)
print("LightGBM R2 Score:", score)
```
