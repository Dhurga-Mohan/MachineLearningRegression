```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

dataset = pd.read_csv("insurance_pre.csv")

dataset.shape
dataset.info()

dataset = pd.get_dummies(dataset, drop_first=True, dtype=int)

independent = dataset[
    ['age', 'bmi', 'children', 'sex_male', 'smoker_yes']
]

dependent = dataset[['charges']]

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    independent,
    dependent,
    test_size=0.30,
    random_state=0
)
```

```python
from sklearn.ensemble import RandomForestRegressor

regressor = RandomForestRegressor(random_state=0)

regressor.fit(X_train, y_train.values.ravel())

y_pred = regressor.predict(X_test)

from sklearn.metrics import r2_score

r_score = r2_score(y_test, y_pred)

print("R2 Score:", r_score)
```

```python
criterion_values = ['squared_error', 'absolute_error']
max_features_values = [10, 100, 'sqrt', 'log2']
n_estimators_values = [10, 100]

results = []

for criterion in criterion_values:
    for max_features in max_features_values:
        for n_estimators in n_estimators_values:

            regressor = RandomForestRegressor(
                criterion=criterion,
                max_features=max_features,
                n_estimators=n_estimators,
                random_state=0
            )

            regressor.fit(X_train, y_train.values.ravel())

            y_pred = regressor.predict(X_test)

            r_score = r2_score(y_test, y_pred)

            results.append({
                'criterion': criterion,
                'max_features': max_features,
                'n_estimators': n_estimators,
                'R2': r_score
            })

results_df = pd.DataFrame(results)

results_df
```

```python
max_depth_values = [None, 3, 5, 8, 10, 15]

results = []

for max_depth in max_depth_values:

    regressor = RandomForestRegressor(
        criterion='squared_error',
        max_features='sqrt',
        n_estimators=100,
        max_depth=max_depth,
        random_state=0
    )

    regressor.fit(X_train, y_train.values.ravel())

    y_pred = regressor.predict(X_test)

    r_score = r2_score(y_test, y_pred)

    results.append({
        'max_depth': max_depth,
        'R2': r_score
    })

results_df = pd.DataFrame(results).sort_values(
    by='R2', ascending=False
)

results_df
```

```python
min_samples_split_values = [2, 5, 10, 20]

results = []

for min_samples_split in min_samples_split_values:

    regressor = RandomForestRegressor(
        criterion='squared_error',
        max_features='sqrt',
        n_estimators=100,
        max_depth=8,
        min_samples_split=min_samples_split,
        random_state=0
    )

    regressor.fit(X_train, y_train.values.ravel())

    y_pred = regressor.predict(X_test)

    r_score = r2_score(y_test, y_pred)

    results.append({
        'min_samples_split': min_samples_split,
        'R2': r_score
    })

results_df = pd.DataFrame(results).sort_values(
    by='R2', ascending=False
)

results_df
```

```python
min_samples_leaf_values = [1, 2, 5, 10]

results = []

for min_samples_leaf in min_samples_leaf_values:

    regressor = RandomForestRegressor(
        criterion='squared_error',
        max_features='sqrt',
        n_estimators=100,
        max_depth=8,
        min_samples_split=2,
        min_samples_leaf=min_samples_leaf,
        random_state=0
    )

    regressor.fit(X_train, y_train.values.ravel())

    y_pred = regressor.predict(X_test)

    r_score = r2_score(y_test, y_pred)

    results.append({
        'min_samples_leaf': min_samples_leaf,
        'R2': r_score
    })

results_df = pd.DataFrame(results).sort_values(
    by='R2', ascending=False
)

results_df
```
